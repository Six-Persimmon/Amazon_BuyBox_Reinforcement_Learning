```python
import numpy as np

import matplotlib.pyplot as plt

# Parameters
t = np.arange(1, 101)
lambda1 = 1
lambda2 = 0.5

# Generate Poisson events at each time point
events1 = np.random.poisson(lam=lambda1, size=len(t))
events2 = np.random.poisson(lam=lambda2, size=len(t))

# Cumulative number of events (N)
cum_events1 = np.cumsum(events1)
cum_events2 = np.cumsum(events2)

# Plot cumulative number of events
plt.figure(figsize=(10, 5))
plt.plot(t, cum_events1, label='λ = 1')
plt.plot(t, cum_events2, label='λ = 0.5')
plt.xlabel('Time')
plt.ylabel('Cumulative Number of Events (N)')
plt.title('Cumulative Number of Events for Two Poisson Processes')
plt.legend()
plt.grid(True)
plt.show()

# Plot number of events at each time point
plt.figure(figsize=(10, 5))
plt.plot(t, events1, label='λ = 1')
plt.plot(t, events2, label='λ = 0.5')
plt.xlabel('Time')
plt.ylabel('Number of Events at Each Time Point')
plt.title('Number of Events per Time Point for Two Poisson Processes')
plt.legend()
plt.grid(True)
plt.show()
```
